# Step 1 Understanding the Basics

## What is MapReduce?
MapReduce is a programming paradigm used in Hadoop for handling big data analysis using multiple machines. There are two phases in which MapReduce operates; the first phase is called a Map phase during which data is processed and transformed while the second one is called a Reduce phase where results are aggregated. In between every phase, MapReduce saves intermediate results to the hard drive which is then loaded into the subsequent phase. The repeated loading and saving of results cause MapReduce to be slow since disk I/O operation is slower compared to the operations conducted in-memory.

## Why is Spark better?
Apache Spark offers a few advantages over MapReduce. Firstly, it operates in memory thus increasing speed. Instead of saving intermediate results to the hard drive, Spark keeps them in memory and only uses hard drive when needed. Memory is several times faster than the hard drive and hence Spark will be much faster than MapReduce. Secondly, it allows easy integration with DataFrames. Spark provides DataFrames which are table-like data structures that offer simple and efficient API for data manipulation such as loading, cleaning, filtering and analysis.

# Step 9 Advance Concepts:
- Wide Transformation
- Narrow Transformation
- Shuffle

There are two categories of transformations depending on how data is transferred among partitions in Spark.

**Narrow Transformation** are those transformation operations in which partitions can be computed individually, without having to refer to data from another partition. Thus, the data remains at its original place. They include filter(), select(), and withColumn(). Narrow transformations are very fast since there is no transfer of data at all.

**Wide Transformation**are those transformation operations in which partitions can be computed individually, without having to refer to data from another partition. Thus, the data remains at its original place. They include filter(), select(), and withColumn(). Narrow transformations are very fast since there is no transfer of data at all.

**Shuffle**: This data transfer among partitions (or across the network within the cluster) is called shuffle. It occurs during wide transformations. Shuffle is the costliest operation in Spark since transferring data between partitions takes some time. Hence, it may slow down the process. Therefore, it is advisable to avoid unneeded wide transformations and shuffles as far as possible.

# Step 10 Creating a Simple Pipeline:    

```python
def run_pipeline(input_path):    
    # Loading Dataset
    df = spark.read.csv(input_path, header=True, inferSchema=True)
    
    # Cleaning the dataset
    df = df.dropDuplicates()
    df = df.withColumn("region", F.initcap(F.trim(F.col("region"))))
    df = df.na.drop(subset=["salary"])
    
    # Transform the dataset
    df = df.withColumnRenamed("experience_years", "exp_years")
    df = df.withColumn("salary", F.col("salary").cast("int"))
    
    # Apply Filter
    df = df.filter(F.col("age") > 30)
    
    # Using Aggregate functions
    result = df.groupBy("department").agg(
        F.count("*").alias("headcount"),
        F.round(F.avg("salary"), 0).alias("avg_salary")
    )
    return result


final_result = run_pipeline("../Data/dataset.csv")
final_result.show()
```
